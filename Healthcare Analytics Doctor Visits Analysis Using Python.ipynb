## 8. Bivariate Analysis

Bivariate analysis involves studying the relationship between two variables at a time. It helps us understand whether changes in one variable are associated with differences in another variable.

In this section, we will analyze the relationships between doctor visits and factors such as age, gender, income, illness, health status, and chronic conditions.

Different visualization techniques such as scatter plots, box plots, line charts, bar charts, and stacked bar charts will be used.

### 8.1 Age vs Doctor Visits

This analysis examines the relationship between a patient's age and the number of doctor visits. It helps us understand whether doctor visits tend to vary across different age groups.

```python
age_visit_analysis = df.groupby('age')['visits'].agg(['count', 'mean']).reset_index()

print("Average Doctor Visits by Age:")
print(age_visit_analysis)
```

### Visualization: Age vs Doctor Visits Scatter Plot

A scatter plot is used to visualize the relationship between age and doctor visits. Each point represents an observation, allowing us to identify patterns, concentration, and variation between the two numerical variables.

```python
sns.scatterplot(
    data=df,
    x='age',
    y='visits',
    color='teal',
    alpha=0.6
)

plt.title('Age vs Doctor Visits', fontsize=16)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Number of Doctor Visits', fontsize=12)

plt.tight_layout()
plt.show()
```

### Observation

The scatter plot shows how doctor visits are distributed across different ages. The points indicate variation in the number of visits among patients of different ages. The overall pattern can be used to examine whether age and doctor visits show an apparent relationship in the dataset.

### 8.2 Gender vs Doctor Visits

This analysis examines the relationship between gender and the number of doctor visits. It helps us compare the distribution of doctor visits across different gender categories.

```python
gender_visit_analysis = df.groupby('gender')['visits'].agg(['count', 'mean' , 'median']).reset_index()
print("Doctor Visit Statistics by Gender:")
print(gender_visit_analysis)
```

### Visualization: Gender vs Doctor Visits Box Plot

A box plot is used to compare the distribution of doctor visits across gender categories. It shows the median, spread, and possible outliers in the number of visits for each category.

```python
plt.figure(figsize=(11,6))
sns.boxplot(
    data=df,
    x='gender',
    y='visits',
    palette='Set2'
)
plt.title('Doctor Visits by Gender',fontsize=16)
plt.xlabel('Gender',fontsize=12)
plt.ylabel('Number of Doctor Visits',fontsize=12)
plt.tight_layout()
plt.show()
```

### Observation

The box plot compares the distribution of doctor visits across gender categories. It shows differences in the median, spread, and range of visits between the groups and can also highlight observations with unusually high or low visit counts.

### 8.3 Income vs Doctor Visits

This analysis examines the relationship between patient income and the number of doctor visits. It helps us understand how healthcare utilization varies across different income levels.

```python
df['income_group'] = pd.qcut(
    df['income'],
    q=4,
    labels=['Low', 'Lower-Middle', 'Upper-Middle', 'High'],
    duplicates='drop'
)

income_visit_analysis = df.groupby('income_group', observed=True)['visits'].agg(
    ['count', 'mean', 'median']
).reset_index()

print("Doctor Visit Statistics by Income Group:")
print(income_visit_analysis)
```

### Visualization: Income vs Doctor Visits Scatter Plot

A scatter plot is used to visualize the relationship between income and doctor visits. The points represent individual patient records, while the trend line helps identify the overall direction of the relationship.

```python
plt.figure(figsize=(11, 6))

sns.regplot(
    data=df,
    x='income',
    y='visits',
    scatter_kws={'alpha': 0.5},
    line_kws={'linewidth': 2}
)

plt.title('Income vs Doctor Visits', fontsize=16)
plt.xlabel('Income', fontsize=12)
plt.ylabel('Number of Doctor Visits', fontsize=12)

plt.tight_layout()
plt.show()
```

### Observation

The scatter plot shows the distribution of doctor visits across different income levels. The trend line provides an overall indication of the relationship between income and doctor visits, while the scattered points show variation among individual patients.

### 8.4 Illness vs Doctor Visits

This analysis examines the relationship between the number of illnesses and the number of doctor visits. It helps us understand whether doctor visits vary across different illness levels.

```python
illness_visit_analysis = df.groupby('illness')['visits'].agg(
    ['count', 'mean', 'median']
).reset_index()

print("Doctor Visit Statistics by Illness Level:")
print(illness_visit_analysis)
```

### Visualization: Illness vs Doctor Visits Line Chart

A line chart is used to visualize the average number of doctor visits across different illness levels. The line makes it easier to observe the pattern or trend in average visits as the illness level changes.

```python
plt.figure(figsize=(11, 6))

plt.plot(
    illness_visit_analysis['illness'],
    illness_visit_analysis['mean'],
    marker='o',
    linewidth=2,
    color='purple'
)

plt.title('Average Doctor Visits by Illness Level', fontsize=16)
plt.xlabel('Illness Level', fontsize=12)
plt.ylabel('Average Doctor Visits', fontsize=12)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Observation

The line chart shows how the average number of doctor visits changes across different illness levels. The pattern can be used to examine whether patients with different levels of illness tend to have different levels of healthcare utilization.

### 8.5 Health vs Doctor Visits

This analysis examines the relationship between a patient's health status and the number of doctor visits. It helps us compare the average healthcare utilization across different health status categories.

```python
health_visit_analysis=df.groupby('health')['visits'].agg(['count','mean','median']).reset_index()
print("Doctor Visit Statistics by Health Status:")
print(health_visit_analysis)
```

### Visualization: Health Status vs Doctor Visits Bar Chart

A bar chart is used to compare the average number of doctor visits across different health status categories. The height of each bar represents the average number of visits for that health status.

```python
plt.figure(figsize=(11, 6))

sns.barplot(
    data=health_visit_analysis,
    x='health',
    y='mean',
    color='cornflowerblue'
)

plt.title('Average Doctor Visits by Health Status', fontsize=16)
plt.xlabel('Health Status', fontsize=12)
plt.ylabel('Average Doctor Visits', fontsize=12)

plt.tight_layout()
plt.show()
```

### Observation

The bar chart compares the average number of doctor visits across different health status categories. The differences between the bars show how average healthcare utilization varies among patients with different recorded health statuses.

### 8.6 Chronic Condition vs Doctor Visits

This analysis examines the relationship between chronic health conditions and the number of doctor visits. The dataset contains information about new chronic conditions (`nchronic`) and long-term chronic conditions (`lchronic`). This analysis compares doctor visit patterns across these chronic-condition categories.

```python
chronic_visit_analysis = df.groupby(
    ['nchronic', 'lchronic']
)['visits'].agg(['count', 'mean', 'median']).reset_index()

print("Doctor Visit Statistics by Chronic Conditions:")
print(chronic_visit_analysis)
```

### Visualization: Chronic Conditions vs Doctor Visits Stacked Bar Chart

A stacked bar chart is used to compare the average number of doctor visits across combinations of new and long-term chronic conditions. This visualization helps show how doctor visits vary across different chronic-condition categories.

```python
chronic_pivot = chronic_visit_analysis.pivot(
    index='nchronic',
    columns='lchronic',
    values='mean'
)

ax = chronic_pivot.plot(
    kind='bar',
    stacked=True,
    figsize=(11, 6),
    colormap='viridis'
)

plt.title('Average Doctor Visits by Chronic Conditions', fontsize=16)
plt.xlabel('New Chronic Condition', fontsize=12)
plt.ylabel('Average Doctor Visits', fontsize=12)
plt.legend(title='Long-Term Chronic Condition')

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Observation

The stacked bar chart compares average doctor visits across different combinations of new and long-term chronic conditions. The chart helps identify variations in healthcare utilization among patients with different chronic-condition categories.

## 9. Multivariate Analysis

Multivariate analysis involves analyzing three or more variables together to understand relationships and patterns within the dataset.

In this section, we will use a correlation matrix, heatmap, and pair plot to examine the relationships between important numerical healthcare variables such as doctor visits, age, income, illness, reduced activity, and health status.

### 9.1 Correlation Matrix

A correlation matrix is used to measure the strength and direction of relationships between numerical variables. The correlation coefficient ranges from -1 to +1.

A positive value indicates a positive relationship, while a negative value indicates a negative relationship. A value close to zero indicates a weak or little linear relationship.

```python
numerical_columns = df.select_dtypes(include=np.number).columns

correlation_matrix = df[numerical_columns].corr()

print("Correlation Matrix:")
print(correlation_matrix.round(2))
```

### Visualization: Correlation Matrix

The correlation matrix presents the pairwise correlation values between the numerical variables in the dataset. It provides a numerical view of how the variables are related to each other.

```python
correlation_matrix.round(2)
```

### Observation

The correlation matrix shows the strength and direction of relationships among the numerical variables. Values closer to +1 indicate a stronger positive relationship, while values closer to -1 indicate a stronger negative relationship. Values near 0 indicate a weaker linear relationship.

### 9.2 Correlation Heatmap

A correlation heatmap is used to visually represent the correlation between numerical variables. Different colors show the strength and direction of relationships, making it easier to identify positive, negative, and weak correlations.

```python
numerical_columns = df.select_dtypes(include=np.number).columns
correlation_matrix = df[numerical_columns].corr()

print("Correlation matrix calculated successfully.")
```

### Visualization: Correlation Heatmap

The heatmap provides a visual representation of the correlation matrix. The correlation values are displayed inside the cells, making it easier to compare the relationships between numerical variables.

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5,
    square=True
)

plt.title('Correlation Heatmap of Healthcare Variables', fontsize=16)
plt.xlabel('Variables', fontsize=12)
plt.ylabel('Variables', fontsize=12)
plt.tight_layout()
plt.show()
```

### Observation

The correlation heatmap shows the strength and direction of relationships between the numerical variables. Darker shades represent stronger relationships, while lighter shades indicate weaker relationships. The numerical values in the cells help identify the direction and strength of each correlation.

### 9.3 Pair Plot

A pair plot is used to visualize the relationships between multiple numerical variables at the same time. It displays scatter plots for pairs of variables and distribution plots along the diagonal.

This helps us identify patterns, trends, and possible relationships among important healthcare variables.

```python
pairplot_columns = ['visits', 'age', 'income', 'illness', 'reduced', 'health']

print("Variables selected for Pair Plot:")
print(pairplot_columns)
```

### Visualization: Pair Plot of Healthcare Variables

The pair plot shows the relationships between doctor visits, age, income, illness, reduced activity, and health status. The scatter plots show relationships between pairs of variables, while the diagonal plots show the distribution of each variable.

```python
g = sns.pairplot(
    df[pairplot_columns],
    diag_kind='hist'
)

g.fig.set_size_inches(14, 12)
g.fig.suptitle('Pair Plot of Healthcare Variables', y=1.02, fontsize=16)

plt.show()
```

### Observation

The pair plot provides an overall view of relationships among multiple healthcare variables. It helps identify patterns, clusters, and variations between variables and can be used to examine whether some variables show noticeable relationships with doctor visits.

## 10. Data Visualization

Data visualization is used to present important patterns and insights from the healthcare dataset in a graphical form.

In this section, selected variables such as doctor visits, age, illness, and health status are visualized together to provide an overall view of healthcare utilization patterns.

A bar chart is used to compare the average values of selected numerical variables. This provides a simple visual summary of important characteristics in the healthcare dataset.

```python
summary_values = {
    'Doctor Visits': avg_visits,
    'Age': avg_age,
    'Illness': avg_illness,
    'Health': avg_health
}

plt.figure(figsize=(12, 7))

bars = plt.bar(
    summary_values.keys(),
    summary_values.values(),
    width=0.6,
    color='teal'
)

plt.title('Healthcare Dataset Summary', fontsize=17)
plt.xlabel('Variables', fontsize=12)
plt.ylabel('Average Value', fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{bar.get_height():.2f}',
        ha='center',
        va='bottom',
        fontsize=11
    )

plt.tight_layout()
plt.show()
```

The project involved data loading, data cleaning, data preparation, descriptive statistics, univariate analysis, bivariate analysis, multivariate analysis, and data visualization using Python.

Different visualization techniques such as histograms, count plots, bar charts, pie charts, box plots, scatter plots, line charts, stacked bar charts, heatmaps, and pair plots were used to understand patterns and relationships within the dataset.

Overall, the analysis demonstrates how Python-based data analytics can be used to explore healthcare data, identify patterns in doctor visits, and understand relationships between patient characteristics and healthcare utilization.

The project provides a foundation for further healthcare analytics and can be extended with predictive modeling and machine learning techniques in the future.
